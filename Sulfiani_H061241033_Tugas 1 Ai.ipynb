## **1. STUDI KASUS GPR**

### **Identifikasi Pipa Bawah Permukaan**

#### **— Role**

Anda berperan sebagai seorang ahli geofisika eksplorasi yang memiliki keahlian dalam metode Ground Penetrating Radar (GPR), pengolahan radargram, dan interpretasi objek bawah permukaan.


#### **— Task**

Analisis studi kasus untuk mengidentifikasi keberadaan pipa bawah permukaan menggunakan metode Ground Penetrating Radar (GPR). Interpretasikan data radargram dan tentukan posisi serta perkiraan kedalaman pipa berdasarkan data yang tersedia.


####  **— Context**

Sebuah area kampus akan digunakan untuk pembangunan saluran drainase. Sebelum dilakukan penggalian, perlu diketahui keberadaan utilitas bawah tanah seperti pipa agar tidak terjadi kerusakan. Survei GPR dilakukan karena metode ini bersifat non-destruktif dan mampu mendeteksi perubahan sifat elektromagnetik bawah permukaan.

Objek seperti pipa dapat menghasilkan pola refleksi berbentuk hiperbola pada radargram. Namun, pola hiperbola tidak selalu menunjukkan pipa sehingga interpretasi harus mempertimbangkan karakteristik data dan kondisi lokasi.


#### **— Input**


Gunakan data berikut:

Survei GPR menggunakan antena 500 MHz dengan panjang lintasan 50 meter. Pada radargram ditemukan anomali hiperbola pada posisi lintasan sekitar 18 meter dengan two-way travel time 12 ns. Kecepatan gelombang elektromagnetik diperkirakan 0,10 m/ns.

Anomali kedua ditemukan pada posisi sekitar 34 meter dengan two-way travel time 18 ns. Kondisi permukaan relatif datar dan sebagian area berupa perkerasan.


#### **— Constraints**


Gunakan prinsip propagasi gelombang elektromagnetik dalam GPR. Jika menghitung kedalaman, gunakan two-way travel time dan kecepatan gelombang yang diberikan. Jangan membuat data tambahan.

Jangan langsung menyatakan bahwa semua anomali hiperbola merupakan pipa. Jelaskan kemungkinan interpretasi berdasarkan karakteristik radargram dan berikan alasan geofisika yang mendukung.


####  **— Output**


Berikan hasil analisis hanya dalam bentuk narasi ilmiah. Jelaskan permasalahan, prinsip GPR, karakteristik radargram, anomali yang ditemukan, posisi dan perkiraan kedalaman target, kemungkinan objek yang menyebabkan anomali, dasar interpretasi, serta kesimpulan. Jangan menggunakan tabel, bullet point, numbering, atau daftar.

## **2. STUDI KASUS SEISMIK REFRAKSI**

### **Menentukan Ketebalan Lapisan dan Kedalaman Batuan Dasar**

#### **— Role**


Anda berperan sebagai seorang ahli geofisika yang memiliki keahlian dalam metode seismik refraksi, analisis travel time, pembuatan kurva waktu-tempuh, dan interpretasi struktur bawah permukaan.


#### **— Task**


Analisis data seismik refraksi untuk menentukan model lapisan bawah permukaan, kecepatan gelombang setiap lapisan, serta perkiraan kedalaman batas antar lapisan.

#### **— Context**


Sebuah kawasan akan digunakan untuk pembangunan infrastruktur. Informasi mengenai ketebalan tanah pelapukan dan kedalaman batuan dasar diperlukan untuk mengetahui kondisi bawah permukaan.

Metode seismik refraksi digunakan dengan menghasilkan gelombang seismik di permukaan dan merekam waktu kedatangan gelombang menggunakan geophone. Perbedaan waktu tiba gelombang digunakan untuk menentukan kecepatan lapisan bawah permukaan. Lapisan yang lebih dalam umumnya dapat memiliki kecepatan gelombang lebih tinggi dibandingkan lapisan permukaan.


#### **— Input**


Survei menggunakan 12 geophone dengan interval 5 meter dan satu sumber getaran pada posisi awal lintasan. Hasil pengukuran menunjukkan adanya perubahan kemiringan pada kurva travel time.

Data interpretasi awal menunjukkan gelombang langsung memiliki kecepatan sekitar 400 m/s, sedangkan gelombang refraksi pertama memiliki kecepatan sekitar 1.200 m/s. Gelombang refraksi kedua memiliki kecepatan sekitar 2.500 m/s.

Kurva travel time menunjukkan adanya tiga lapisan dengan perubahan slope yang cukup jelas.


#### **— Constraints**


Gunakan prinsip dasar seismik refraksi dan hubungan antara slope kurva travel time dengan kecepatan gelombang.

Jangan membuat jumlah lapisan tambahan apabila tidak didukung oleh data. Jelaskan bahwa kedalaman lapisan merupakan hasil interpretasi berdasarkan data travel time. Bedakan antara data pengukuran dan hasil interpretasi.

Gunakan satuan yang konsisten dan jangan membuat nilai waktu tiba yang tidak diberikan.

#### **— Output**


Berikan hasil analisis hanya dalam bentuk narasi ilmiah. Jelaskan prinsip metode seismik refraksi, karakteristik data travel time, interpretasi jumlah lapisan, kecepatan masing-masing lapisan, perubahan kecepatan, kemungkinan kondisi material bawah permukaan, perkiraan kedalaman batas lapisan jika dapat ditentukan, serta kesimpulan mengenai kondisi bawah permukaan. Jangan menggunakan tabel, bullet point, numbering, atau daftar.

## **3. STUDI KASUS RESISTIVITAS LISTRIK / ERT**

### **Identifikasi Zona Air Tanah**

#### **— Role**

Anda berperan sebagai seorang ahli geofisika hidrogeofisika yang memiliki keahlian dalam metode Electrical Resistivity Tomography (ERT), konfigurasi elektroda, inversi resistivitas, dan interpretasi kondisi bawah permukaan.


####  **— Task**


Analisis data Electrical Resistivity Tomography (ERT) untuk mengidentifikasi kemungkinan zona yang berhubungan dengan air tanah dan membedakannya dari material bawah permukaan lainnya berdasarkan nilai resistivitas.


#### **— Context**


Sebuah wilayah mengalami kesulitan memperoleh sumber air bersih. Survei geofisika dilakukan untuk mengetahui kondisi bawah permukaan dan memperkirakan keberadaan zona yang memiliki kemungkinan mengandung air.

Metode ERT menginjeksikan arus listrik ke dalam tanah melalui elektroda dan mengukur beda potensial yang dihasilkan. Data tersebut digunakan untuk memperoleh distribusi resistivitas bawah permukaan. Material dengan kandungan air yang lebih tinggi dapat memiliki resistivitas yang berbeda dibandingkan material kering, tetapi nilai resistivitas juga dipengaruhi oleh jenis batuan, porositas, salinitas, dan kandungan lempung.


#### **— Input**


Survei menggunakan konfigurasi Wenner dengan 24 elektroda dan jarak antar elektroda 5 meter.

Hasil inversi resistivitas menunjukkan tiga zona utama. Zona dekat permukaan memiliki resistivitas sekitar 80–150 ohm meter. Pada kedalaman sekitar 10–20 meter ditemukan zona dengan resistivitas 15–30 ohm meter yang relatif menerus secara lateral. Di bawah zona tersebut ditemukan material dengan resistivitas sekitar 200–500 ohm meter.

Wilayah dengan resistivitas rendah berada pada bagian tengah lintasan dan memiliki ketebalan relatif lebih besar dibandingkan zona di sekitarnya.


#### **— Constraints**


Interpretasikan nilai resistivitas berdasarkan prinsip metode ERT dan karakteristik material bawah permukaan.

Jangan langsung menyatakan zona resistivitas rendah sebagai air tanah karena resistivitas rendah juga dapat disebabkan oleh material lempung atau air dengan kandungan ion tinggi.

Bandingkan kemungkinan interpretasi berdasarkan data yang tersedia. Jangan membuat data tambahan mengenai muka air tanah atau litologi yang tidak diberikan.

#### **— Output**

Berikan hasil analisis hanya dalam bentuk narasi ilmiah. Jelaskan prinsip ERT, konfigurasi pengukuran, distribusi resistivitas, karakteristik zona resistivitas rendah dan tinggi, kemungkinan material penyusunnya, kemungkinan hubungan zona resistivitas rendah dengan air tanah, faktor yang dapat menyebabkan ketidakpastian, serta kesimpulan interpretasi bawah permukaan. Jangan menggunakan tabel, bullet point, numbering, atau daftar.

## **4. STUDI KASUS METODE MAGNETIK**

### **Identifikasi Struktur Geologi dan Zona Intrusi**

#### **— Role**


Anda berperan sebagai seorang ahli geofisika eksplorasi yang memiliki keahlian dalam metode geomagnetik, analisis anomali medan magnet, koreksi data magnetik, pembuatan peta anomali, dan interpretasi struktur geologi bawah permukaan.

#### **— Task**


Analisis data geomagnetik untuk mengidentifikasi kemungkinan struktur geologi bawah permukaan dan zona batuan yang memiliki kontras sifat kemagnetan.

#### **— Context**


Suatu wilayah penelitian diduga memiliki struktur geologi berupa sesar dan kemungkinan keberadaan batuan dengan sifat magnetik yang berbeda dari batuan di sekitarnya.

Metode geomagnetik digunakan untuk mengukur variasi medan magnet bumi yang dipengaruhi oleh sifat kemagnetan batuan bawah permukaan. Perbedaan susceptibilitas magnetik dapat menghasilkan anomali magnetik yang dapat digunakan untuk membantu mengidentifikasi kontak batuan, struktur geologi, atau tubuh batuan tertentu.


#### **— Input**


Survei geomagnetik dilakukan pada area seluas 2 km × 2 km dengan jarak antar titik pengukuran 50 meter.

Setelah dilakukan koreksi variasi harian dan koreksi medan magnet bumi, diperoleh peta anomali magnetik. Pada bagian barat daya ditemukan anomali positif dengan amplitudo sekitar 250 nT. Di bagian tengah wilayah terdapat zona gradien magnetik yang cukup tajam dengan perubahan sekitar 180 nT dalam jarak kurang dari 200 meter.

Bagian timur wilayah menunjukkan anomali magnetik yang relatif rendah dan lebih homogen.

#### **— Constraints**


Interpretasikan anomali magnetik berdasarkan prinsip geomagnetik dan kontras sifat magnetik batuan.

Jangan langsung menyatakan anomali positif sebagai jenis batuan tertentu tanpa informasi pendukung. Zona gradien magnetik dapat mengindikasikan perubahan lateral sifat magnetik atau kemungkinan struktur geologi, tetapi interpretasinya harus tetap mempertimbangkan kemungkinan lain.

Jangan membuat kedalaman sumber anomali jika tidak terdapat data atau metode perhitungan yang mendukung.

#### **— Output**


Berikan hasil analisis hanya dalam bentuk narasi ilmiah. Jelaskan prinsip metode geomagnetik, karakteristik anomali magnetik, keberadaan anomali positif dan negatif, zona gradien magnetik, kemungkinan perubahan sifat batuan, kemungkinan keberadaan struktur geologi, dasar interpretasi, keterbatasan data, dan kesimpulan mengenai kondisi bawah permukaan. Jangan menggunakan tabel, bullet point, numbering, atau daftar.

## **5. STUDI KASUS METODE GRAVITASI**

### **Identifikasi Cekungan Sedimen dan Struktur Bawah Permukaan**

#### **— Role**

Anda berperan sebagai seorang ahli geofisika yang memiliki keahlian dalam metode gravitasi, analisis anomali Bouguer, interpretasi densitas batuan, dan pemodelan struktur bawah permukaan.

#### **— Task**


Analisis data gravitasi untuk mengetahui variasi densitas bawah permukaan dan mengidentifikasi kemungkinan keberadaan cekungan sedimen atau struktur geologi berdasarkan pola anomali gravitasi.


#### **— Context**


Suatu wilayah diduga merupakan bagian dari cekungan sedimen. Informasi mengenai konfigurasi bawah permukaan diperlukan untuk memahami struktur geologi wilayah tersebut.

Metode gravitasi memanfaatkan variasi percepatan gravitasi akibat perbedaan densitas material bawah permukaan. Material dengan densitas lebih rendah dapat menghasilkan anomali gravitasi relatif rendah, sedangkan material dengan densitas lebih tinggi dapat menghasilkan anomali relatif tinggi.

Data gravitasi perlu melalui beberapa koreksi sebelum digunakan untuk interpretasi, seperti koreksi drift, lintang, elevasi, dan koreksi Bouguer sesuai kebutuhan survei.


#### **— Input**


Survei gravitasi dilakukan pada area seluas 5 km × 5 km dengan jarak antar titik pengukuran sekitar 250 meter.

Setelah dilakukan koreksi dan diperoleh anomali Bouguer, bagian tengah wilayah menunjukkan anomali relatif rendah dengan nilai sekitar -35 mGal. Nilai anomali meningkat secara bertahap menuju bagian utara dan selatan hingga sekitar -10 mGal.

Pada bagian timur terdapat zona perubahan anomali yang relatif tajam dalam jarak sekitar 1 kilometer.


#### **— Constraints**


Interpretasikan anomali gravitasi berdasarkan prinsip hubungan antara densitas material dan medan gravitasi.

Jangan menentukan jenis batuan tertentu hanya berdasarkan nilai anomali gravitasi tanpa informasi pendukung. Anomali rendah dapat mengindikasikan material berdensitas relatif rendah, tetapi penyebabnya dapat lebih dari satu.

Jangan menentukan kedalaman atau ketebalan cekungan secara pasti apabila tidak terdapat data pemodelan atau parameter tambahan yang mendukung.

Bedakan antara pengamatan pola anomali dan interpretasi geologi.


#### **— Output**


Berikan hasil analisis hanya dalam bentuk narasi ilmiah. Jelaskan prinsip metode gravitasi, proses umum pengolahan data, karakteristik anomali Bouguer, pola anomali rendah dan tinggi, kemungkinan variasi densitas bawah permukaan, kemungkinan keberadaan cekungan sedimen atau struktur geologi, keterbatasan interpretasi, serta kesimpulan berdasarkan data yang tersedia. Jangan menggunakan tabel, bullet point, numbering, atau daftar.
